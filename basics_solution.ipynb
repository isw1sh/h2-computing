## Linear search

**Linear search** — check each element in turn until the target is found or the end of the list is reached. Works on unsorted data. Time complexity O(n). *(Exam appearances: hash table searches 2016/2018/2025 fall back to linear probing.)*

```python
# Linear search
def linear_search(values, target):
    # check each element in turn; return its position, or -1 if not found
    for index in range(len(values)):
        if values[index] == target:
            return index
    return -1

# test main program
data = [34, 7, 91, 15, 62, 7, 48]
print(linear_search(data, 15))     # 3
print(linear_search(data, 7))      # 1 (first occurrence)
print(linear_search(data, 99))     # -1 (not in the list)
```

## Binary search (iterative)

**Binary search (iterative)** — repeatedly halve the search range of a **sorted** list by comparing the target with the middle element. Time complexity O(log n).

```python
# Binary search (iterative)
def binary_search(values, target):
    # values must be sorted into ascending order
    low = 0
    high = len(values) - 1
    while low <= high:
        middle = (low + high) // 2
        if values[middle] == target:
            return middle
        elif values[middle] < target:
            low = middle + 1
        else:
            high = middle - 1
    return -1

# test main program
data = [3, 8, 15, 21, 34, 48, 62, 91]
print(binary_search(data, 21))     # 3
print(binary_search(data, 3))      # 0 (first element)
print(binary_search(data, 91))     # 7 (last element)
print(binary_search(data, 50))     # -1 (not in the list)
```

## Binary search (recursive)

**Binary search (recursive)** — the same halving idea expressed recursively: the base cases are an empty range (not found) or the middle element matching. *(Examined in 2022 Task 2.4 and the mock prelim Task 3.3.)*

```python
# Binary search (recursive)
def binary_search_recursive(values, target, low, high):
    # base case: the range is empty, so the target is not in the list
    if low > high:
        return -1
    middle = (low + high) // 2
    if values[middle] == target:
        return middle
    elif values[middle] < target:
        return binary_search_recursive(values, target, middle + 1, high)
    else:
        return binary_search_recursive(values, target, low, middle - 1)

# test main program
data = [3, 8, 15, 21, 34, 48, 62, 91]
print(binary_search_recursive(data, 34, 0, len(data) - 1))   # 4
print(binary_search_recursive(data, 8, 0, len(data) - 1))    # 1
print(binary_search_recursive(data, 99, 0, len(data) - 1))   # -1
```

## Bubble sort

**Bubble sort** — repeatedly step through the list, swapping neighbouring values that are out of order; after each pass the largest remaining value has 'bubbled' to the end. Time complexity O(n²). *(Examined 2022 Task 2.2.)*

```python
# Bubble sort
def bubble_sort(values):
    # after each pass, the largest remaining value is in its final position
    for last in range(len(values) - 1, 0, -1):
        for index in range(last):
            if values[index] > values[index + 1]:
                # swap the neighbouring values that are out of order
                values[index], values[index + 1] = values[index + 1], values[index]
    return values

# test main program
data = [56, 25, 4, 98, 0, 18, 4, 5, 7, 0]
print(bubble_sort(data))           # [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

## Insertion sort

**Insertion sort** — grow a sorted section at the front of the list; each new value is slid left past larger values into its correct place. Time complexity O(n²), but fast on nearly-sorted data. *(Examined 2021 Task 2.2; no-builtin sort in 2024 Task 1.3.)*

```python
# Insertion sort
def insertion_sort(values):
    # grow a sorted section at the start of the list
    for index in range(1, len(values)):
        current = values[index]
        position = index - 1
        # shift larger values one place to the right
        while position >= 0 and values[position] > current:
            values[position + 1] = values[position]
            position = position - 1
        values[position + 1] = current
    return values

# test main program
data = [56, 25, 4, 98, 0, 18, 4, 5, 7, 0]
print(insertion_sort(data))        # [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

## Quicksort

**Quicksort** — choose a pivot, partition the list into values smaller and larger than the pivot, and recursively sort each partition. Average time complexity O(n log n); worst case O(n²) on already-sorted data with a poor pivot. *(Examined 2018 Task 2.1 and 2021 Task 2.3.)*

```python
# Quick sort
def quick_sort(values):
    # base case: a list of 0 or 1 items is already sorted
    if len(values) <= 1:
        return values
    # partition around the first element as the pivot
    pivot = values[0]
    smaller = [value for value in values[1:] if value < pivot]
    larger = [value for value in values[1:] if value >= pivot]
    # recursively quicksort each partition either side of the pivot
    return quick_sort(smaller) + [pivot] + quick_sort(larger)

# test main program
data = [56, 25, 4, 98, 0, 18, 4, 5, 7, 0]
print(quick_sort(data))            # [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

## Merge sort

**Merge sort** — recursively split the list in half until single elements remain, then merge the sorted halves back together. Time complexity O(n log n) in every case. *(Examined 2020 Task 2.2 and 2022 Task 2.3.)*

```python
# Merge sort
def merge(left, right):
    # merge two sorted lists into one sorted list
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i = i + 1
        else:
            merged.append(right[j])
            j = j + 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

def merge_sort(values):
    # base case: a list of 0 or 1 items is already sorted
    if len(values) <= 1:
        return values
    middle = len(values) // 2
    return merge(merge_sort(values[:middle]), merge_sort(values[middle:]))

# test main program
data = [56, 25, 4, 98, 0, 18, 4, 5, 7, 0]
print(merge_sort(data))            # [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

## Linked list

**Linked list** — a chain of nodes, each holding a data value and a reference to the next node. Order is given by the links, not by position in memory. Insertion at the head is O(1). *(Examined 2017 Task 3 and 2021 Task 3; mock prelim Task 2.)*

```python
# Linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, value):
        # insert at the beginning (head) of the list
        node = Node(value)
        node.next = self.head
        self.head = node

    def delete(self, value):
        # remove the first node holding value; return True if removed
        previous = None
        current = self.head
        while current is not None:
            if current.data == value:
                if previous is None:
                    self.head = current.next
                else:
                    previous.next = current.next
                return True
            previous = current
            current = current.next
        return False

    def search(self, value):
        current = self.head
        while current is not None:
            if current.data == value:
                return True
            current = current.next
        return False

    def count(self):
        number = 0
        current = self.head
        while current is not None:
            number = number + 1
            current = current.next
        return number

    def to_string(self):
        items = []
        current = self.head
        while current is not None:
            items.append(str(current.data))
            current = current.next
        return "[" + ", ".join(items) + "]"

# test main program
linked_list = LinkedList()
for value in [5, 7, 1, 3]:
    linked_list.insert(value)
print(linked_list.to_string())     # [3, 1, 7, 5]
print(linked_list.search(7))       # True
print(linked_list.delete(1))       # True
print(linked_list.to_string())     # [3, 7, 5]
print(linked_list.count())         # 3
```

## Stack

**Stack** — Last In, First Out (LIFO). Items are pushed onto and popped off the top. Used for reversing, bracket matching and base conversion. *(Examined 2019 Task 4 and 2025 Task 1.)*

```python
# Stack
class Stack:
    def __init__(self):
        self.items = []          # 1-dimensional list to hold the stack
        self.top = 0             # index of the next free space

    def push(self, item):
        # if the stack is full, increase the length of the list
        if self.top == len(self.items):
            self.items.append(None)
        self.items[self.top] = item
        self.top = self.top + 1

    def pop(self):
        # remove and return the top item, or None if the stack is empty
        if self.top == 0:
            return None
        self.top = self.top - 1
        return self.items[self.top]

    def size(self):
        return self.top

# test main program: reverse a word using the LIFO property
stack = Stack()
for character in "COMPUTING":
    stack.push(character)
reversed_word = ""
while stack.size() > 0:
    reversed_word = reversed_word + stack.pop()
print(reversed_word)               # GNITUPMOC
print(stack.pop())                 # None (the stack is now empty)
```

## Queue

**Queue** — First In, First Out (FIFO). Items join at the tail and leave from the head. The circular version reuses freed spaces by wrapping the pointers with MOD. *(Examined 2021 Task 3.3 and, as a circular queue, 2023 Task 2.)*

```python
# Queue (circular)
class Queue:
    def __init__(self, size):
        self.items = [None] * size   # fixed-size list for the circular queue
        self.head = 0                # index of the first item
        self.tail = 0                # index of the next free space
        self.number = 0              # number of items in the queue

    def enqueue(self, item):
        # insert at the tail; return False if the queue is full
        if self.number == len(self.items):
            return False
        self.items[self.tail] = item
        self.tail = (self.tail + 1) % len(self.items)
        self.number = self.number + 1
        return True

    def dequeue(self):
        # remove and return the head item, or None if the queue is empty
        if self.number == 0:
            return None
        item = self.items[self.head]
        self.head = (self.head + 1) % len(self.items)
        self.number = self.number - 1
        return item

# test main program
queue = Queue(4)
for item in ["A", "B", "C", "D"]:
    print("enqueue", item, ":", queue.enqueue(item))   # all True
print("enqueue E :", queue.enqueue("E"))               # False - queue is full
print(queue.dequeue())            # A (first in, first out)
print(queue.dequeue())            # B
print("enqueue E :", queue.enqueue("E"))               # True - space wrapped around
print(queue.dequeue())            # C
```

## Binary search tree

**Binary search tree (BST)** — each node has a left and right child; smaller values go left, larger (or equal) go right. An in-order traversal (left, node, right) visits the values in ascending order; post-order visits children before the node. *(Examined 2016 Task 3, 2022 Task 3 and 2024 Task 2.)*

```python
# Binary search tree
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def insert(self, value):
        # recursive: smaller values go left, larger or equal go right
        if value < self.data:
            if self.left is None:
                self.left = TreeNode(value)
            else:
                self.left.insert(value)
        else:
            if self.right is None:
                self.right = TreeNode(value)
            else:
                self.right.insert(value)

    def search(self, value):
        if value == self.data:
            return True
        if value < self.data:
            return self.left is not None and self.left.search(value)
        return self.right is not None and self.right.search(value)

    def in_order(self):
        # left subtree, then the node, then the right subtree (ascending order)
        if self.left is not None:
            self.left.in_order()
        print(self.data, end=" ")
        if self.right is not None:
            self.right.in_order()

    def post_order(self):
        # both subtrees first, then the node
        if self.left is not None:
            self.left.post_order()
        if self.right is not None:
            self.right.post_order()
        print(self.data, end=" ")

# test main program
values = [50, 30, 70, 20, 40, 60, 80]
tree = TreeNode(values[0])
for value in values[1:]:
    tree.insert(value)
tree.in_order()                    # 20 30 40 50 60 70 80
print()
tree.post_order()                  # 20 40 30 60 80 70 50
print()
print(tree.search(60))             # True
print(tree.search(65))             # False
```

## Hash table

**Hash table** — a key is converted to a list index by a hashing function (commonly `key MOD table size`); collisions are resolved by linear probing to the next free slot. Average O(1) insert and search. *(Examined 2016 Task 2, 2018 Task 3, 2020 Task 1 and 2025 Task 4.)*

```python
# Hash table
TABLE_SIZE = 11
hash_table = [None] * TABLE_SIZE

def calculate_hash(key):
    # hash(k) = k mod s, where s is the size of the table
    return key % TABLE_SIZE

def insert(key, data):
    # store [key, data] at the hashed index, using linear probing
    index = calculate_hash(key)
    for _ in range(TABLE_SIZE):
        if hash_table[index] is None:
            hash_table[index] = [key, data]
            return True
        index = (index + 1) % TABLE_SIZE
    return False                    # the table is full

def find(key):
    # probe from the hashed index; None means the key is not stored
    index = calculate_hash(key)
    for _ in range(TABLE_SIZE):
        record = hash_table[index]
        if record is None:
            return None
        if record[0] == key:
            return record[1]
        index = (index + 1) % TABLE_SIZE
    return None

# test main program
insert(23, "red")                  # 23 mod 11 = 1
insert(34, "green")                # 34 mod 11 = 1 -> collision, probes to 2
insert(45, "blue")                 # 45 mod 11 = 1 -> collision, probes to 3
insert(7, "yellow")                # 7 mod 11 = 7
for index in range(TABLE_SIZE):
    print(index, hash_table[index])
print(find(34))                    # green (found by probing)
print(find(99))                    # None (not stored)
```
