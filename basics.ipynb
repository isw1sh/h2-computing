## Linear search

**Linear search** — check each element in turn until the target is found or the end of the list is reached. Works on unsorted data. Time complexity O(n). *(Exam appearances: hash table searches 2016/2018/2025 fall back to linear probing.)*

```python
# Linear search
def linear_search(values,target):
    for i in range(len(values)):
        if values[i] == target:
            return i 
    return 'not found'

data = [34, 7, 91, 15, 62, 7, 48]
print(linear_search(data, 15))     # 3
print(linear_search(data, 7))      # 1 (first occurrence)
print(linear_search(data, 99))     # -1 (not in the list)
```

## Binary search (iterative)

**Binary search (iterative)** — repeatedly halve the search range of a **sorted** list by comparing the target with the middle element. Time complexity O(log n).

```python
# Binary search (iterative)
def binary_search(values,target):
    low = 0
    high = len(values) - 1
    while low <= high:
        mid = (low + high) // 2
        if values[mid] == target:
            return mid
        elif values[mid] < target:
            low = mid + 1            
        else:
            high = mid - 1
            
    return 'not in list'

# test main program
data = [3, 8, 15, 21, 34, 48, 62, 91]
print(binary_search(data, 21))     # 3
print(binary_search(data, 3))      # 0 (first element)
print(binary_search(data, 91))     # 7 (last element)
print(binary_search(data, 50))     # -1 (not in the list)
```

## Binary search (recursive)

**Binary search (recursive)** — the same halving idea expressed recursively: the base cases are an empty range (not found) or the middle element matching. *(Examined in 2022 Task 2.4 and the mock prelim Task 3.3.)*

```python
# Binary search (recursive)
def binary_search_recursive(values,target, low, high):
    if low > high:
        return 'not in list'
    mid = (low + high) // 2
    if values[mid] == target:
        return mid 
    elif values[mid] < target:
        return binary_search_recursive(values,target,mid+1, high)
    else:
        return binary_search_recursive(values,target, low, mid-1)  
    
data = [3, 8, 15, 21, 34, 48, 62, 91]
print(binary_search_recursive(data, 34, 0, len(data) - 1))   # 4
print(binary_search_recursive(data, 8, 0, len(data) - 1))    # 1
print(binary_search_recursive(data, 99, 0, len(data) - 1))   # -1  
```

## Bubble sort

**Bubble sort** — repeatedly step through the list, swapping neighbouring values that are out of order; after each pass the largest remaining value has 'bubbled' to the end. Time complexity O(n²). *(Examined 2022 Task 2.2.)*

```python
# Bubble sort
def bubble_sort(values):
    for i in range(len(values) - 1):    
        swapped = False
        for j in range(len(values) - 1 - i):
            if values[j] > values[j+1]:
                values[j] , values [j+1] = values[j+1] , values[j]
                swapped = True
        if not swapped:
            break
    return values

data = [56, 25, 4, 98, 0, 18, 4, 5, 7, 0]
print(bubble_sort(data)) 
                            
```
