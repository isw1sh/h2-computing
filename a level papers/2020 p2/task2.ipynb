**Task 2.1**

Write a function `task2_1(filename, quantity, maximum)` that:

- accepts three parameters:
  - `filename`, a string representing the name of a file
  - `quantity`, an integer representing the number of random integers to generate
  - `maximum`, representing the largest value that a random integer can take
- generates `quantity` random numbers between 0 and `maximum` (inclusive)
- writes those values, one per line, to a file named `filename`. **[4]**

Generate 1000 random numbers between 0 and 5000 (inclusive) and save them to a file called
`randomnumbers_<your name>_<centre number>_<index number>.txt` **[2]**


````python
# Task 2.1
from random import *
def task2_1(filename, quantity, maximum):
    with open(filename , 'w') as file:
        for i in range(quantity):
            number = randint(1,maximum)
            number = str(number)
            file.write(number + '\n')
    print('File created')

task2_1('randomnumbers.txt' , 1000 , 5000)
````

`task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98]`

should return `True`. **[2]**


````python
    # Task 2.2
    def merge(left,right):
        merged = []
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                merged.append(left[i])
                i = i + 1
            else:
                merged.append(right[j])
                j = j + 1

        merged += left[i:]
        merged += right[j:]
        return merged


    def task2_2(list_of_integers):
        if len(list_of_integers) <= 1:
            return list_of_integers
        
        mid = len(list_of_integers) // 2
        left = list_of_integers[:mid]
        right = list_of_integers[mid:]

        return merge((task2_2(left)) , task2_2(right))

    print(task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98])
````

````python
# Task 2.3
def task2_3(filename_in , filename_out):
    list_of_integers = []
    with open(filename_in, 'r') as file:
        for line in file:
            list_of_integers.append(int(line))
    sorted_list = task2_2(list_of_integers)

    with open(filename_out , 'w') as file:
        for num in sorted_list:
            file.write(str(num) + '\n')

task2_3('randomnumbers.txt' , 'sortednumbers.txt')
````
